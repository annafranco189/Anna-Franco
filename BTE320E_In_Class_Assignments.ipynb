```python
sales = float(input("Enter Sales Value ($):"))
tax_rate = float(input("Enter tax rate (%):"))

cogs = 0.4 * sales
gross_profit = sales - cogs
taxes = gross_profit * tax_rate / 100
net = gross_profit - taxes


print(f"Sales: {sales}$")
print(f"Cost of Goods Sold: {cogs}$")
print(f"Gross Profit: {gross_profit}$")
print(f"Taxes: {taxes}$")
print(f"Net Income: {net}$")

```

```python
flatfee = 5
hourly = 2.5
print("Hours|Total Fee")

hours = 1
while hours < 9:
  fee = flatfee + (hours*hourly)
  if fee > 20:
    print(f"   {hours} | 20")
  elif fee < 10:
    print(f"   {hours} | 10")
  else:
    print(f"   {hours} | {fee}")
  hours += 1
```

```python
import random

choice = input("Do you want to play? (yes/no): ").lower()

while choice == "yes":
  player_move = input("Enter your move (rock, paper, scissors): ").lower()
  computer_move = random.choice(["rock","paper","scissors"])

  if player_move == computer_move:
      print("draw")
  elif (
      (player_move == "rock" and computer_move == "scissors") or
      (player_move == "paper" and computer_move == "rock") or
      (player_move == "scissors" and computer_move == "paper")
  ):
    print("you win!")
  elif (
      (player_move == "rock" and computer_move == "paper") or
      (player_move == "paper" and computer_move == "scissors") or
      (player_move == "scissors" and computer_move == "rock")):
    print("you loose")
  else:
    print("invalid move")

  choice = input("Do you want to play again? (yes/no): ").lower()

if choice == "no":
  print("Thanks for playing")
```

```python
class Greeting:
  def __init__(self, first, last):
    self.first = first
    self.last = last

  def display(self, formal=False):
    if formal:
      print(f"{self.last}, {self.first}")
    else:
      print(f"{self.first}")

  def __str__(self):
    return "Instance of Class Greeting"

name = Greeting("Bob", "Tom")
print(name)
name.display()
name.display(formal=True)
```

```python
def getValues(n):
    assert isinstance(n, int), "'n' must be an integer"
    items = []

    for i in range(n):
        items.append(float(input("Enter a value:")))

    if len(items) < 2:
        raise ValueError("the length of list 'items' cannot be < 2")
    return items

def calcAverage(alist):
    assert isinstance(alist, list), "'alist' must be a list"
    try:
        return sum(alist) / len(alist)
    except Exception as e:
        print(f"Error: {e}")


def main():
    n = int(input("Enter a number of items to be added to the list: "))
    alist = getValues(n)
    avg = calcAverage(alist)
    print(f"Average weight: {avg} lbs.")

main()
```

```python
class Person:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name

    def display(self, formal=False):
        if formal:
            print(f"{self.last_name}, {self.first_name}")
        else:
            print(self.first_name)

    def __str__(self):
        return f"Instance of class {self.__class__.__name__}"

name = Person ("John" , "Doe")
print(name)
name.display()
name.display(formal=True)
```

```python
class Employee:
    def __init__(self,name,wage,hours):
      self.name = name
      self.__hourly_wage = wage
      self.__hours = hours

    def get_wages(self):
      return self.__hourly_wage

    def set_wage(self, newWage):
      if newWage <= 0:
        print("Invalid wage")
      else:
        self.__hourly_wage = newWage

    def get_hours(self):
      return self.__hours

    def calculate_pay(self):
      return self.__hourly_wage * self.__hours

    def display_info(self):
      print(f"Employee Name: {self.name}\nHourly Wages: ${self.__hourly_wage}")

class Manager(Employee):
    def __init__(self,name,wage,hours,bonus):
      super().__init__(name,wage,hours)
      self.bonus = bonus

    def calculate_pay(self):
      return (1 + self.bonus) * self.get_wages() * self.get_hours()

employee = Employee("Bob", 50, 10)
manager = Manager("Alice", 100, 15, 0.2)
print(employee)
print(manager)
print()
print(f"Employee {employee.name} weekly salary is: ${employee.calculate_pay()}")
print(f"Manager {manager.name} weekly salary is: ${manager.calculate_pay()} that includes a {manager.bonus * 100}% bonus")
```
